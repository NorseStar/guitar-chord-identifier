# First look at a recording

Load one A-chord recording and plot its waveform. The horizontal axis shows time; the vertical axis shows amplitude. This helps us see the quiet start and how the sound fades. This early example uses a Test recording for a visual check only. Later experiments use Train recordings to choose settings.

```python
import librosa.display
import matplotlib.pyplot as plt

audio, sr = librosa.load("../data/raw/data/Test/A/A_acoustic_guitar_fender_fa_series_test_1.wav")

plt.figure(figsize=(12,4))
librosa.display.waveshow(audio, sr=sr)
plt.show()
```
